### Convert datatype object into datetime

```python
# fear_greed_index dataset
df1['date'] = pd.to_datetime(df1['date'], errors = 'coerce')

# historical_data dataset
df2['Timestamp IST'] = pd.to_datetime(df2['Timestamp IST'], errors = 'coerce')
```

### Extracting Date

```python
# fear_greed_index dataset
df1['Date'] = df1['date'].dt.date

# historical_data dataset
df2['Date'] = df2['Timestamp IST'].dt.date

# Extracting Time 
df2['Time'] = df2['Timestamp IST'].dt.time
```

### Merge Both Datasets

```python
# Merged both dataset by date
merged_df = pd.merge(df2, df1, on = 'Date', how = 'inner')
```

```python
# Checking the columns after merge
merged_df.columns
```

### Drop Irrelevent Columns

```python
merged_df.drop(columns = ['Timestamp IST','Transaction Hash','Timestamp','timestamp','date',], axis = 1, inplace = True)
```

### handle columns name representation

```python
merged_df.columns = merged_df.columns.str.replace(' ','_')
```

## Finding KPIs

### Daily PnL per Trader

```python
daily_pnl = merged_df.groupby(['Account','Date'])['Closed_PnL'].sum().reset_index(name = 'Profit/loss')
print('Daily Profit and Loss per trader is:\n',daily_pnl)
```

### Win Rate

```python
wining_trades = merged_df[merged_df['Closed_PnL']>0.0].shape[0]
total_trades = merged_df['Closed_PnL'].count()

wining_rate = (wining_trades/total_trades)*100
print('Wining rate:', wining_rate)
```

### Average Trade Size

```python
avg_trade_size = merged_df.groupby(['classification'])['Size_USD'].mean()
print('Average Trade size by classification is:\n',avg_trade_size)
```

### Leverage Distribution

```python
merged_df.describe()
```

### Number of Trades Per Day

```python
trade_per_day = merged_df.groupby(['Date'])['Trade_ID'].count().reset_index(name='Number_of_Trades')
print('Number of trade per day is:\n',trade_per_day)
```

### Long/Short Ratio

```python
merged_df['Direction'].value_counts()
```

```python
short = merged_df[merged_df['Direction'] == 'Open Short'].shape[0]
long = merged_df[merged_df['Direction'] == 'Open Long'].shape[0]
ratio = long/short
print('The ratio of long and short is:',ratio)
```

### Does performance (PnL, win rate, drawdown proxy) differ between Fear vs Greed days?

```python
# PnL analyssis on the basis of Fear and Greed
pnl_analysis = merged_df.groupby('classification')['Closed_PnL'].agg(['mean','sum','std'])
print('PnL Analysis on the basis of the greed and fear is:\n',pnl_analysis.loc[['Greed', 'Fear']])
```

```python
# Win Rate Analysis on the basis of Fear and Greed
merged_df['win'] = merged_df['Closed_PnL']>0

win_rate = merged_df.groupby(['classification'])['win'].mean()*100
print('Win rate on the basis of Greed and Fear is:\n',win_rate.loc[['Greed','Fear']])
```

```python
# Drawdown Proxy
downtown_proxy = merged_df.groupby(['classification'])['Closed_PnL'].agg(Drawdown_Proxy = 'std')
print(downtown_proxy.loc[['Greed','Fear']])
```

### Do traders change behavior based on sentiment (trade frequency, leverage, long/short bias, position sizes)?

```python
# trade frequency
trade_frequency = merged_df.groupby(['classification'])['Account'].count().reset_index(name = 'trade_frequency')
print(trade_frequency)
```

```python
# long/short bias
bias = pd.crosstab(merged_df['classification'], merged_df['Side'])
print(bias)
```

```python
# position sizes
avg_position_size = merged_df.groupby(['classification'])['Size_USD'].mean().reset_index(name = 'avg_position_size')
print(avg_position_size)
```

### High Exposure vs Low Exposure Traders

```python
exposer = merged_df.groupby(['Account'])['Size_USD'].mean()
median_exposer = exposer.median()
exposer_segment = exposer.apply(lambda x : 'High Exposer' if x>median_exposer else 'Low Exposer').reset_index(name = 'Exposer Type')
print(exposer_segment)
```

### frequent vs infrequent traders

```python
trade_count = merged_df.groupby(['Account']).size()
median_trade = trade_count.median()
trade_frequency = trade_count.apply(lambda x : 'Frequent Trader' if x > median_trade else 'Infrequent Trader').reset_index(name = 'Trader frequency')
print(trade_frequency)
```

### consistent winners vs inconsistent traders

```python
pnl_std = merged_df.groupby(['Account'])['Closed_PnL'].std()
median_pnl = pnl_std.median()
trader_segment = pnl_std.apply(lambda x : 'Consistant Trader' if x < median_pnl else 'Inconsistant Trader').reset_index(name = 'Trader Cnsistency')
print(trader_segment)
```

## Visualization

```python
# PnL Distribution
sns.barplot(x = 'classification', y = 'Closed_PnL', data = merged_df, hue = 'classification', palette = 'coolwarm')
plt.title('Profit/Loss by Market Sentiment')
plt.xlabel('Classification of Sentiment')
plt.ylabel('Profit/Loss')
plt.show()
```

```python
# Short/Long bias
bias = pd.crosstab(merged_df['classification'], merged_df['Side'])
sns.heatmap(bias, annot=True)
plt.title('Buy and Selling Behavior by Market Sentiment')
plt.show()
```

```python
# Trade Frequency
sns.countplot(x = 'classification', data = merged_df, hue = 'Side', palette = 'Spectral')
plt.title('Trade Frequency by Market Sentiment')
plt.xlabel('Market Sentiment')
plt.ylabel('Number of traders')
plt.show()
```
